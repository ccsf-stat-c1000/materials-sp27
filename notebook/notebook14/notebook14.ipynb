##(i). Load GoogleSheet Data into Python

In class, I will discuss how the dataset is structured for the ANOVA computations in Python.   


```python
# Import pandas.
import pandas as pd

# Copy/paste your shareable link for Blood Pressure dataset, modify it, read in and check file.
filepath = 'Blood Pressure Data Link'

# name the data frame "bloodpressure_df"
bp_df = pd.read_csv(filepath)

# Enter the code to read the header of the data set.
bp_df.head()
```

## (ii) Creating side-by-side boxplots by treatment method

It is good habit to first look at your dataset in boxplots before performing ANOVA.


This will not be tested on the final exam.


```python
# Import matplotlib.pyplot as plt
import matplotlib.pyplot as plt

# Create side-by-side boxplots for each treatment method
fig = plt.figure(figsize= (5, 5))
ax = fig.add_subplot(111)

data = [bp_df['BPValue'][bp_df['Treatment'] == 'Medication'],
        bp_df['BPValue'][bp_df['Treatment'] == 'Exercise'],
        bp_df['BPValue'][bp_df['Treatment'] == 'Diet']]

ax.boxplot(data,
           labels= ['Medication', 'Exercise', 'Diet'])

plt.xlabel("Treatment Method")
plt.ylabel("Blood Pressure Reduction Value")

plt.show()
```

## (iii) Creating an ANOVA table


`statsmodels` is a Python package that provides a complement to `scipy` for statistical computations. We will use code from this library to create an ANOVA table.

**This will be needed for the final exam.**


```python
# Import libraries
import statsmodels.api as sm
from statsmodels.formula.api import ols

# Run one-way ANOVA
bpmodel = ols('BPValue ~ C(Treatment)', data=bp_df).fit()

# Print ANOVA table
print(sm.stats.anova_lm(bpmodel, typ=2))
```

## Question 5 -- Type Data into Pyton

**alternative method**

```python
import pandas as pd
import statsmodels.api as sm
from statsmodels.formula.api import ols

# Create sample DataFrame
data = {'BPValue': [9, 10, 11, 11, 9, 8, 9, 7, 10, 9, 5, 8, 9, 7, 9],
        'Treatment': ['M', 'M', 'M', 'M', 'M', 'E', 'E', 'E', 'E', 'E', 'D', 'D', 'D', 'D', 'D']}
BP = pd.DataFrame(data)

# Fit Ordinary Least Squares (OLS) model
model = ols('BPValue ~ C(Treatment)', data=BP).fit()
# Perform ANOVA
anova_table = sm.stats.anova_lm(model, typ=2)
print(anova_table)
```

## Computing Sample Means and SDs by Treatment Variable

This part will not be on the final exam. But you'd need to know how to create the ANOVA table for the final exam.

```python
# Import pandas.
import pandas as pd

# Copy/paste your shareable link for tv show dataset, modify it, read in and check file.
filepath = 'TV dataset'

# name the data frame and check the head of the data frame
tv = pd.read_csv(filepath)
tv.head()

# complute the sample mean and SD by program
SSmean = tv[tv['Program'] == 'Sesame Street']['Serotonin'].mean()
SSsd = tv[tv['Program'] == 'Sesame Street']['Serotonin'].std()

print('Sesame Street mean = ', SSmean)
print('Sesame Street sd = ', SSsd)

Fmean = tv[tv['Program'] == 'Futurama']['Serotonin'].mean()
Fsd = tv[tv['Program'] == 'Futurama']['Serotonin'].std()

print('Futurama mean = ', Fmean)
print('Futurama sd = ', Fsd)

Bmean = tv[tv['Program'] == 'BBC News']['Serotonin'].mean()
Bsd = tv[tv['Program'] == 'BBC News']['Serotonin'].std()

print('BBC News mean = ', Bmean)
print('BBC News sd = ', Bsd)

Nmean = tv[tv['Program'] == 'No Program']['Serotonin'].mean()
Nsd = tv[tv['Program'] == 'No Program']['Serotonin'].std()

print('No Program mean = ', Nmean)
print('No Program sd = ', Nsd)
```
