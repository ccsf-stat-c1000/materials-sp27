## Handout 6.4 Enter the Values in Python Code

Example #1 - Dependent Samples

```python
# Import pandas.
import pandas as pd

# dataset for Handout 6.4
zinc = pd.DataFrame({'bottom': [0.43,0.266,0.567,0.531,0.707,0.716,0.651,0.589,0.469,0.723], 'surface': [0.415,0.238,0.390,0.410,0.605,0.609,0.632,0.523,0.411,0.612]})

# add a new column and call it 'diff'
zinc['diff'] = zinc['bottom'] - zinc['surface']

# The code below will show you the dataframe with the new column added.
print(zinc)

# The code below computes and prints the mean and SD for 'diff'.
x_bar = zinc['diff'].mean()
sd = zinc['diff'].std()

print('sample mean = ', x_bar)
print('sample SD = ', sd)
```

Example 2 - Dependent Samples

```python
# Import pandas.
import pandas as pd

# dataset for Handout 6.4
twins = pd.DataFrame({'A': [87,88,72,88,85,85,76,93,76,81,79,72], 'B': [83,96,80,76,79,98,84,89,82,80,88,85]})

# add a new column and call it 'diff'
twins['diff'] = twins['A'] - twins['B']

# The code below will show you the dataframe with the new column added.
print(twins)

# The code below computes and prints the mean and SD for 'diff'.
x_bar = twins['diff'].mean()
sd = twins['diff'].std()

print('sample mean = ', x_bar)
print('sample SD = ', sd)
```

Example 3 - Independent Samples

```python
# Import pandas.
import pandas as pd

# dataset for Handout 6.4
children = pd.DataFrame({'sugar': [40.3, 29.7, 45.7, 30.3, 44.2, 37.4, 33.6, 23.8, 30.3, 28.4]})
adults = pd.DataFrame({'sugar': [30.2, 26.5, 34.5, 28.5, 20.7, 18.4, 21.4, 27.5]})

# The code below will show you the dataframe for children's cereals.
print(children)

# The code below computes and prints the mean and SD for children's cereals.
x_barC = children['sugar'].mean()
sdC = children['sugar'].std()

print('children sample mean = ', x_barC)
print('children sample SD = ', sdC)

# The code below will show you the dataframe for adults' cereals.
print(adults)

# The code below computes and prints the mean and SD for adults' cereals.
x_barA = adults['sugar'].mean()
sdA = adults['sugar'].std()

print('adult ample mean = ', x_barA)
print('adult sample SD = ', sdA)
```
