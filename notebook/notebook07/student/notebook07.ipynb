# 1. Creating Contingency Tables

In Pandas, there is a function called `crosstab()` that we can use to create a contingency table.

Here's the code to create a contingency table:

`pd.crosstab(df['row name'], df['column name'], margins = True')`

#2. Computing Percentages: Row, Column, or Cell Percents

You can use the normalize argument within the Pandas `crosstab()` function to create a contingency table that shows percentage values instead of counts:

*`normalize = 'all'` : shows percentage relative to all values.

*`normalize = 'index'` : shows percentage as total of row values.

*`normalize = 'columns'` : shows percentage as total of column values.


### <font color="red">➡</font> Run the Cell   

<font color='red'>Exercise 3</font>  

Run the cell below to see the cell percentages for the contingency table.


```python
# run the code below.
cont_table3 = pd.crosstab(soda_df['Drink Soda'], soda_df['Age Group'], margins = True, normalize = 'all')
print(cont_table3)
```

#3. Creating Stacked Bar Graphs

You can create a stacked bar plot in pandas by using the built-in function `.plot(kind='bar', stacked=True)`


### <font color="red">➡</font> Run the Cell
