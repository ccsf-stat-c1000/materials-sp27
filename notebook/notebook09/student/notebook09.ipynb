```python
# create a new row where YearsExperience is 1 and Salary is 200000
new_row = pd.DataFrame({"YearsExperience":[1], "Salary":[200000]})

# use the function concat to add the new row to salary data frame and call it new_salary_df
new_salary_df=pd.concat([salary_df, new_row])

# this plots the graph using the new_salary_df
sns.regplot(data = new_salary_df, x = 'YearsExperience', y = 'Salary', ci = None)

# this call gets the regression statistics
slope, intercept, r_value, p_value, std_err = stats.linregress (new_salary_df['YearsExperience'], new_salary_df['Salary'])

# print out the items we care about
print('slope = ', slope)
print('y-intercept', intercept)
print('r_value', r_value)
```
