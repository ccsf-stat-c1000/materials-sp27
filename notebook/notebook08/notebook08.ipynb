# Question #1: Footlength vs. Height

**Using a GoogleSheet dataset**

Seaborn has a function that creates a scatterplot and the code is `.regplot`, and the code is given below:

`sns.regplot(data = df, x = 'column 1', y ='column 2', ci = None`

We will also need the SciPy library. SciPy is an open-source software for mathematics, science, and engineering. It includes many functions for statistics, and higher level mathematics such as optimization, integration, linear algebra, Fourier transforms, Differential Equation solvers, and more. We will use the `stats.linregress` function from SciPy here.

### <font color="red">➡</font> Run the Cell

```python
# Import libaries.
import pandas as pd
import seaborn as sns
from scipy import stats

filepath = 'Footlength and Height link'
# Name the dataframe 'footlen' for the foot length vs height dataset.
footlen= pd.read_csv(filepath)
print(footlen.head())

# Create the scatterplot
sns.regplot(data = footlen, x = 'Foot Length', y = 'Height', ci = None)

# Compute and print correlation coefficient, and slope and y-intercept for the best-fit line.

slope, intercept, r_value, p_value, std_err = stats.linregress (footlen['Foot Length'], footlen['Height'])
print('slope = ', slope)
print('y-intercept', intercept)
print('r_value', r_value)
```
