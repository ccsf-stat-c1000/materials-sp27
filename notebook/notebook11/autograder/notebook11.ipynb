# Question #1: Gettysburg Address


Enter the values in Python code to create a dataframe and name it Getty, then compute the sample mean.

### <font color="red">➡</font> Enter Values and Run the Cell

```python
# Import pandas.
import pandas as pd

# Enter your data for the number of letters in a word for the Gettysburg Address.
# Name the data frame Getty.
Getty = pd.DataFrame({'N_letters': [1, 2, 3]})

# print the data frame to see your data
print(Getty)

# Compute the mean for number letters in a word
x_bar = Getty['N_letters'].mean()

# Print the sample mean
print('sample mean = ', x_bar)
```

#Question #4: Polar Bears


### <font color="red">➡</font> Enter the Values and Run the Cell

```python
# Import pandas.
import pandas as pd

# Enter your data and name the data frame Bears.
Bears = pd.DataFrame({'weight': [1, 2, 3]})

# print the data frame to see your data
print(Bears)

# Compute the mean and SD for weight
x_bar = Bears['weight'].mean()
sd = Bears['weight'].std()

# Print the sample mean and SD
print('sample mean weight = ', x_bar)
print('sample SD weight = ', sd)
```

```python
# Import Pandas
import pandas as pd

filepath = 'https://docs.google.com/spreadsheets/d/1-j1vR6OcrMpsxM-oCrZaR67bih5VOV5tBFSoETJxeJ4/export?format=csv'
# Name the dataframe 'Text' for the textbook cost dataset and print the head of the data frame
Text = pd.read_csv(filepath)
print(Text.head())

# compute and print mean and SD of cost
x_bar = Text['cost'].mean()
sd = Text['cost'].std()

print('sample mean cost = ', x_bar)
print('sample SD cost = ', sd)
```
