## 2. Creating a dot plot  

### What is NumPy?

NumPy stands for Numerical Python. It's a library that performs numerical computations on datasets. To import NumPy, use the keyword `import` followed by the library name, `numpy`. It is also standard to use an abbreviation when importing libraries; for NumPy, this abbreviation is `np`. Putting this all together, we can import Numpy with the following line of code:

      import numpy as np   

 ### What is Matplotlib?  

There are *many* Python libraries designed for data visualization purposes, and here we'll use `matplotlib`. To import Matplotlib, use the keyword `import` followed by the library name, `matplotlib.pyplot`. It's often abbreviated as `plt`. Putting it together, we have the line of code:  

      import matplotlib.pyplot as plt

The code below creates a dotplot using the test data file `Testdata` for the column named `score`.

### <font color="red">➡</font> Edit and Run the Cell

<font color='red'>Exercise 2</font>    

Enter the code to import NumPy on line 2, and the code to import Matplotlib on line 4, and then run the cell to see the dot plot.

```python
# Import NumPy

# Import Matplotlib

# The code below create a dot plot for the column named 'score'.
values, counts = np.unique(test_df['score'], return_counts=True)
fig, ax = plt.subplots(figsize=(10,3))
for value, count in zip(values, counts):
  ax.plot([value]*count, list(range(1,count+1)), 'ro', ms=10, linestyle='')
```
