The code on lines 5 and 8 give you examples of how to enter values into Python. `df1` has one column and `df2` has two columns. Of course you can change the column names -- for example, you can change `Column A` to `age`.



```python
# Import pandas.
import pandas as pd

# Enter your data below, for one column of data.
df1 = pd.DataFrame({'Column A': [0, 10, 10, 10]})

# below is the code for two columns of data
#df2 = pd.DataFrame({'Column A': [1, 2, 9], 'Column B':[3, 4, 5]})

# The code below will show you the data you entered in lines 5 and 8.
print(df1)
#print(df2)

# The code below computes and prints the mean and median for Column A.
x_bar = df1['Column A'].mean()
#median = df1['Column A'].median()

# The code below computes and prints SD for Column A.
sd = df1['Column A'].std()

print('sample mean = ', x_bar)
#print('sample median = ', median)
print('sample SD = ', sd)
```

```python
# Import pandas.
import pandas as pd

# Copy/paste your shareable link for the dataset, modify it, read in and check file.
filepath = 'Golden State Warriors salary link'
# Name the dataframe "GSW_df" for Golden State Warriors salary
GSW_df = pd.read_csv(filepath)

# The code below reads and prints the header of the data set.
print(GSW_df.head())

# Compute and print the sample mean and median for GS Warriors salaries.
GSW_mean = GSW_df['Salary'].mean()
GSW_median = GSW_df['Salary'].median()

print('sample mean =', GSW_mean)
print('sample median =', GSW_median)
```

## 2b. Sample Code for Computing the Mean, Median, and SD of a SUBSET of the Dataset -- Handouts 2.4 & 2.5

This dataset has two columns -- one for City and another for temperature. The City column has two different cities -- San Francisco and St. Louis.

On line 5, create a variable called `filepath` or something similar, and assign it the GoogleSheet file by copying/pasting your shareable link to replace the words `Temperature link`, and then change the last part of the link name from **edit?usp=sharing** to **export?format=csv**

Run the code block below to see the entire dataset.


```python
# Import pandas.
import pandas as pd

# Copy/paste your shareable link for dataset, modify it, read in and check file.
filepath = 'Temperature link'
# Name the dataframe "temp_df" for the temperature dataset.
Temp_df = pd.read_csv(filepath)

# The code below reads and prints the entire temperature dataset.
print(Temp_df.head())

# Compute and print the sample mean and median for San Francisco temperatures.
SFmean = Temp_df[Temp_df['City'] == 'SF']['Temp'].mean()
SFmedian = Temp_df[Temp_df['City'] == 'SF']['Temp'].median()
SFsd = Temp_df[Temp_df['City'] == 'SF']['Temp'].std()

print('SF Mean Temp = ', SFmean)
print('SF Median Temp = ', SFmedian)
print('SF Standard Deviation = ', SFsd)
```
